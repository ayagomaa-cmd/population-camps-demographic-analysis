```python
# second step
# Convert Date to datetime object
df['Date'] = pd.to_datetime(df['Date'])

df['Check_Year'] = df['Date'].dt.year
df['Check_Month'] = df['Date'].dt.month
df['Check_Day'] = df['Date'].dt.day
df[['Date', 'Check_Year', 'Check_Month', 'Check_Day']].head()
```

```python
date_str = df['Date'].astype(str) #slicing

df['Check_Year'] = date_str.str[:4].astype(int)
df['Check_Month'] = date_str.str[5:7].astype(int)
df['Check_Day'] = date_str.str[8:10].astype(int)
```

```python
df['Tent'] = df['Tent'].fillna(0)
```

```python
df['Country'] = df['Country'].astype(str).str.strip()
df['City'] = df['City'].astype(str).str.strip()
```

```python
structural_cols = ['Date', 'Country', 'City', 'PlaceName', 'Houses', 'Members', 'Male', 'Female']

df[structural_cols].isnull().sum()
```

```python
# Save DataFrame to an Excel file
output_file = 'cleaned_dataset.xlsx'
df.to_excel(output_file, index=False)

print(f"File saved successfully as {output_file}!")
```

```python
# third step
# male precentage
import numpy as np
df['Male_Percentage'] = np.where(  #to calculate the percentages ,
                                   #avoid division-by-zero errors
    df['Members'] == 0,
    0,
    (df['Male'] / df['Members']) * 100
)
# female precentage
df['Female_Percentage'] = np.where(
    df['Members'] == 0,
    0,
    (df['Female'] / df['Members']) * 100
)
df[[ 'Male_Percentage', 'Female_Percentage']]
```

```python
#  Average People Per House
df['Avg_People_Per_House'] = np.where(  #(defaults to 0 if Houses == 0)
    df['Houses'] == 0,
    0,
    df['Members'] / df['Houses']
)
df['Avg_People_Per_House']
```

```python
# dominant housing
# housing options
housing_cols = ['Room', 'Flat', 'House', 'Tent']

df[housing_cols].fillna(0)
df['Dominant_Housing'] = df[housing_cols].idxmax(axis=1)

df['Dominant_Housing']
```

```python
# forth step
#columns to sum up
sum = ['Members', 'Male', 'Female', 'Room', 'Flat', 'House', 'Tent']

# Group by 'Country'
df.groupby('Country')[sum].sum().reset_index()
df

```

```python
# temporal trend matrix
df['Date'] = pd.to_datetime(df['Date'])
df['Year'] = df['Date'].dt.year
df.groupby(['Country', 'Year']).sum(numeric_only=True).reset_index()
df
```

```python
# crosstabulation layout
# pivot table
house_distribution_pivot = pd.pivot_table(
    df,
    values='Houses',       #  Total number of houses
    index='Country',       # Rows will be the Countries
    columns='Year',        # Columns will be the Years
    aggfunc='sum',         # total sum (default is average/mean)
    margins=True,          # the margins/sub-totals for both rows and columns
    margins_name='Total',  # Names the margin row/column 'Total' instead of 'All'
    fill_value=0           # Replaces any empty cells (NaN) with 0
)

house_distribution_pivot
```

```python
# fifth step
from sklearn.preprocessing import StandardScaler, MinMaxScaler
# Isolate the continuous numerical features
cont_cols = ['Houses', 'Members', 'Avg_People_Per_House']
df_cont = df[cont_cols]

df_cont
```

```python

x = MinMaxScaler()
scaled_cols = [col + '_scaled' for col in cont_cols]
df[scaled_cols] = x.fit_transform(df[cont_cols])
df[cont_cols + scaled_cols]
```

```python
from sklearn.preprocessing import StandardScaler
x = StandardScaler()
df['Members_Z_Score'] = x.fit_transform(df[['Members']])
outliers = df[df['Members_Z_Score'].abs() > 3]
outliers[['Country', 'City', 'Members', 'Members_Z_Score']]
```

```python
#sixth step
#Generate a grouped bar chart plotting the cumulative population ( Members )
#allocated across different Countries
import matplotlib.pyplot as plt
import seaborn as sns
x = df.groupby('Country')['Members'].sum().reset_index()
x = x.sort_values(by='Members', ascending=False)
sns.set_style('whitegrid')
fig, ax = plt.subplots(figsize=(12, 8))


sns.barplot(
    data=x,
    x='Country',
    y='Members',
    hue='Country'
)

ax.set_title('Cumulative Population (Members) Allocated Across Different Countries', fontsize=16)
ax.set_xlabel('Country', fontsize=14)
ax.set_ylabel('Total Cumulative Population (Members)', fontsize=14)

plt.tight_layout()
plt.show()
```

```python
#Time-Series Line Graph: Create a multi-line temporal chart tracking total population trends over the
#years, assigning a unique distinct color/line style to each individual Country .
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.ticker import MaxNLocator

temporal_data = df.groupby(['Year', 'Country'])['Members'].sum().reset_index()
sns.set_style('whitegrid')
fig, ax = plt.subplots(figsize=(12, 7))

#  Generate the Time-Series Line Graph
sns.lineplot(
    data=temporal_data,
    x='Year',
    y='Members',
    hue='Country',
    style='Country',
    markers=True,
    markersize=8,
    linewidth=2
)

# Label
ax.set_title('Total Population Trends Over Time by Country', fontsize=16)
ax.set_xlabel('Year', fontsize=14)
ax.set_ylabel('Total Population (Members)', fontsize=14)

plt.tight_layout()
plt.show()
```

```python
#Plot an overlaying Seaborn Kernel Density Estimate ( KDE Plot )
#comparing the mathematical distribution spreads of Male_Percentage vs. Female_Percentage across
#the entire study sample.
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('whitegrid')
fig, ax = plt.subplots(figsize=(10, 6))

#  Plot the KDE for Male Percentage
sns.kdeplot(
    data=df,
    x='Male_Percentage',
    fill=True,         # Fills the area under the curve
    color='skyblue',
    label='Male %',

)

#  Plot the KDE for Female Percentage overlaid on the same chart
sns.kdeplot(
    data=df,
    x='Female_Percentage',
    fill=True,
    color='lightcoral',
    label='Female %',

)

#  Label
ax.set_title('Density Distribution Profile: Male vs. Female Percentage', fontsize=16)
ax.set_xlabel('Percentage (%)', fontsize=14)
ax.set_ylabel('Density (Frequency)', fontsize=14)


plt.legend(title='Gender', loc='upper right')
plt.tight_layout()
plt.show()
```

```python
#Linear Regression Scatter Plot: Build a scatter visualization positioning Houses on the X-axis and
#Members on the Y-axis. Integrate a linear regression line via sns.regplot , and map the scatter points to
#distinct colors based on Country .
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style('whitegrid')
fig, ax = plt.subplots(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='Houses',
    y='Members',
    hue='Country',
    s=60,           # Size of the scatter points
    ax=ax
)


sns.regplot(
    data=df,
    x='Houses',
    y='Members',
    scatter=False,  #  set scatter=False so it doesn't draw a second set of uncolored points
    color='black',  # Color of the regression line
    line_kws={'linewidth': 2, 'linestyle': '--'}, # Dashed line s
    ax=ax
)

# Label
ax.set_title('Linear Regression: Houses vs. Population (Members)', fontsize=16)
ax.set_xlabel('Total Houses', fontsize=14)
ax.set_ylabel('Total Population (Members)', fontsize=14)

plt.legend(title='Country', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
#Annotated Correlation Heatmap: Generate a Pearson correlation matrix covering all numerical metrics
#( Houses , Members , Male , Female , Room , Flat , House , Tent ). Visualize this matrix via
#sns.heatmap utilizing a balanced diverging color palette (e.g., coolwarm ) with numeric annotations
#enabled.
import matplotlib.pyplot as plt
import seaborn as sns
metrics = ['Houses', 'Members', 'Male', 'Female', 'Room', 'Flat', 'House', 'Tent']

#   the Pearson correlation matrix
corr_matrix = df[metrics].corr()
plt.figure(figsize=(10, 8))

#  the Annotated Heatmap
sns.heatmap(
    corr_matrix,
    annot=True,         # Turns on the numeric annotations inside the boxes
    fmt=".2f",          # Formats the numbers to 2 decimal places for clean reading
    cmap="coolwarm",    # Uses the requested diverging color palette (blue to red)
    vmin=-1,            # Locks the lower end of the color scale to -1 (perfect negative correlation)
    vmax=1,             # Locks the upper end of the color scale to 1 (perfect positive correlation)
    center=0,           # Ensures 0 (no correlation) is exactly in the middle of the color palette
    linewidths=0.5      # Adds a small grid line between boxes for visual clarity
)

plt.title('Pearson Correlation Heatmap of Numerical Metrics', fontsize=16)
plt.tight_layout()
plt.show()
```
